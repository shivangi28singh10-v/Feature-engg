4) How does loss value help in determining whether the model is good or not?

6) How do we handle categorical variables in Machine Learning? What are the common techniques?
- Categorical variables are handled in machine learning by converting them into numerical form using encoding techniques such as label encoding, one-hot encoding, ordinal encoding, binary encoding, and target encoding.
- Label Encoding :Assigns a unique number to each category.
- One-Hot Encoding (Dummy Encoding) :Creates new binary columns for each category.
- Ordinal Encoding : Used when categories have a natural order.
- Target Encoding (Mean Encoding) : Replace categories with the mean of the target variable.
- Frequency Encoding : Replace category with its frequency in the dataset.
- Binary Encoding : Convert categories to binary numbers, then split into bits.
- Embeddings : Used in Deep Learning—categories are mapped to dense vectors.

21) What is feature scaling? How does it help in Machine Learning?
- Feature Scaling is the process of rescaling numerical features in your dataset so that they are on a similar scale or range.
- Helps algorithms converge faster
- Prevents one feature from dominating others due to larger numeric values.
- How Feature Scaling Helps in ML
- 1) Faster Convergence -Gradient descent reaches minimum faster
- 2) Prevents Feature Domination - Large-scale features don’t overshadow small-scale features
- 3) Improves Accuracy - Especially in distance-based models
- 4) Stable Regularization - L1/L2 penalties work properly when features are scaled.
